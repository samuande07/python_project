```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import plotly.graph_objects as go
import plotly.io as pio
import plotly.colors as colors
pio.templates.default="plotly_white"
import numpy as np

```

```python
data['Order Date']=pd.to_datetime(data['Order Date'],errors='coerce')
data['Ship Date']=pd.to_datetime(data['Ship Date'],errors='coerce')
```

```python
data.head()
```

```python
data['Ship Date'].fillna(data['Order Date'],inplace=True)
```

```python
data.head()
```

```python
data['Order Month'] = data['Order Date'].dt.month #har month kitni sales huyi he
data['Order Year'] = data['Order Date'].dt.year
data['Order Day of week'] = data['Order Date'].dt.dayofweek

```

```python
data.head()
```

**Monthly Sales Analysis**

```python
sales_by_month= data.groupby('Order Month')['Sales'].sum().reset_index()
```

```python
sales_by_month['Order Month']=sales_by_month['Order Month'].astype(int)
```

```python
sales_by_month
```

```python
fig = px.line(sales_by_month,
              x='Order Month',
              y= 'Sales',
              title='Monthly Sales Analysis')
fig.show()
```

**Sales by Category**

```python
sales_by_category=data.groupby('Category')['Sales'].sum().reset_index()
```

```python
sales_by_category
```

```python
fig = px.pie(sales_by_category,
             values='Sales',
             names='Category',
             hole=0.5,
             color_discrete_sequence=colors.qualitative.Pastel)


fig.update_layout(title_text='Sales Analysis by Category',title_font=dict(size=24))

fig.show()



```

**Sales Analysis by sub Category**

```python
sales_by_subcategory=data.groupby('Sub-Category')['Sales'].sum().reset_index()
```

```python
sales_by_subcategory
```

```python
fig=px.bar(sales_by_subcategory,x='Sub-Category',y='Sales',title='Sales analysis by subcategory')
```

```python
fig.show()
```

**Monthly Profit Analysis**

```python

profit_by_month=data.groupby('Order Month')['Profit'].sum().reset_index().astype(int)
```

```python
profit_by_month
```

```python
fig=px.line(profit_by_month,x='Order Month',y='Profit',title='Monthly Profit Analysis')
fig.show()
```

**Profit by Category**

```python
profit_by_category = data.groupby('Category')['Profit'].sum().reset_index()
```

```python
profit_by_category
```

```python
fig=px.bar(profit_by_category,x='Category',y='Profit',title='profit by category analysis')
fig.show()
```
